```python
def get_phone_bill(gb):
    # Base plan cost for 15 GB
    base_cost = 100
    plan_limit = 15
    
    # Check if the customer exceeded the 15 GB limit
    if gb <= plan_limit:
        return base_cost
    else:
        # Calculate excess GB
        excess_gb = gb - plan_limit
        # Convert excess GB to MB (1 GB = 1000 MB)
        excess_mb = excess_gb * 1000
        # Calculate cost for excess data ($0.10 per MB)
        excess_cost = excess_mb * 0.10
        # Return total bill
        return base_cost + excess_cost

# Testing the examples provided:
print(f"Bill for 10 GB: ${get_phone_bill(10)}")   # Expected: 100
print(f"Bill for 15.1 GB: ${get_phone_bill(15.1)}") # Expected: 110
```
